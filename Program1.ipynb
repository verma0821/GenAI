```python
# Install gensim
!pip install gensim

# Import libraries
from gensim.models import KeyedVectors
import gensim.downloader as api

# Load pretrained model
model = api.load("glove-wiki-gigaword-100")

# Check if word exists
print("king" in model.key_to_index)

# Word analogy using vector arithmetic
word1 = "king"
word2 = "man"
word3 = "woman"

result_vector = model[word1] - model[word2] + model[word3]
predicted_word = model.most_similar([result_vector], topn=3)

print(predicted_word)

# Using most_similar directly
result = model.most_similar(
    positive=["king", "woman"],
    negative=["man"],
    topn=3
)

print(result)
```
